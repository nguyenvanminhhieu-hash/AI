```python
# Bài 1

# Hàm trộn mảng
def tron_mang(a, b):
    c = []
    min_len = min(len(a), len(b))

    # Lấy min từng phần tử
    for i in range(min_len):
        c.append(min(a[i], b[i]))

    # Thêm phần dư
    if len(a) > min_len:
        c.extend(a[min_len:])
    if len(b) > min_len:
        c.extend(b[min_len:])

    return c

a = [3, 9, 1, 4]
b = [2, 7, 4, 3, 2, 8]

print("Mảng a:", a)
print("Mảng b:", b)

kq = tron_mang(a, b)

print("Mảng kết quả:", kq)
```

```python
# Bài 2:

# a. Tổng tam giác trên của đường chéo phụ
def tong_tam_giac_tren_cheo_phu(a):
    n = len(a)
    s = 0
    for i in range(n):
        for j in range(n):
            if i + j <= n - 1:   # phía trên (kể cả chéo phụ)
                s += a[i][j]
    return s

    a = [[1, -2, 3],
         [4, 5, -6],
         [7, 8, 9]]

print("Ma trận ban đầu:")
for row in a:
    print(row)

print("\nTổng tam giác trên chéo phụ:", tong_tam_giac_tren_cheo_phu(a))


# b. Đổi số âm thành trị tuyệt đối
def doi_am_thanh_duong(a):
    for i in range(len(a)):
        for j in range(len(a[0])):
            if a[i][j] < 0:
                a[i][j] = abs(a[i][j])

doi_am_thanh_duong(a)
print("\nSau khi đổi âm:")
for row in a:
    print(row)

# c. Thay số chẵn bằng x
def thay_chan_bang_x(a, x):
    for i in range(len(a)):
        for j in range(len(a[0])):
            if a[i][j] % 2 == 0:
                a[i][j] = x

thay_chan_bang_x(a, 0)
print("\nSau khi thay chẵn = 0:")
for row in a:
    print(row)

# d. Kiểm tra toàn chẵn
def toan_chan(a):
    for row in a:
        for x in row:
            if x % 2 != 0:
                return False
    return True

print("\nToàn chẵn:", toan_chan(a))

# e. Kiểm tra ma trận đối xứng
def doi_xung(a):
    n = len(a)
    for i in range(n):
        for j in range(n):
            if a[i][j] != a[j][i]:
                return False
    return True

print("Đối xứng:", doi_xung(a))

# f. Đường chéo chính tăng dần
def cheo_chinh_tang(a):
    n = len(a)
    for i in range(n - 1):
        if a[i][i] >= a[i + 1][i + 1]:
            return False
    return True

print("Chéo chính tăng:", cheo_chinh_tang(a))

# g. Xuất tam giác dưới của đường chéo phụ
def xuat_tam_giac_duoi_cheo_phu(a):
    n = len(a)
    for i in range(n):
        for j in range(n):
            if i + j >= n - 1:   # phía dưới (kể cả chéo phụ)
                print(a[i][j], end=" ")
        print()

print("\nTam giác dưới chéo phụ:")
xuat_tam_giac_duoi_cheo_phu(a)

# h. Đường chéo phụ giảm dần
def cheo_phu_giam(a):
    n = len(a)
    for i in range(n - 1):
        if a[i][n-1-i] <= a[i+1][n-2-i]:
            return False
    return True

print("\nChéo phụ giảm:", cheo_phu_giam(a))
```
